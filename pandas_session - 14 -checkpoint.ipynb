```python
ipl_df = ipl_df.dropna( subset = ["winner"]).reset_index(drop = True)
ipl_df.winner.isnull().sum()                      
# specific column mattum mention panna subset use pannuvom and reset_index(drop = True) index index correct'ah 12345 nu continues'ah vara use pannuvom
```

```python
import numpy as np

ipl_df['match_margin_type'] = np.select(
    [
        (ipl_df['result'] == 'runs') & (ipl_df['result_margin'] > 0),
        (ipl_df['result'] == 'wickets') & (ipl_df['result_margin'] > 0)
    ],
    [
        'Runs',
        'Wickets'
    ],
    default='Tie/No Result'
)

print(ipl_df[['result', 'result_margin', 'match_margin_type']].head())
```

```python
team_cities = {
    'Chennai Super Kings': 'Chennai',
    'Mumbai Indians': 'Mumbai',
    'Royal Challengers Bangalore': 'Bengaluru',
    'Royal Challengers Bengaluru': 'Bengaluru',
    'Kolkata Knight Riders': 'Kolkata',
    'Sunrisers Hyderabad': 'Hyderabad',
    'Rajasthan Royals': 'Jaipur'
}

ipl_df['team1_home_city'] = ipl_df['team1'].map(team_cities)
ipl_df['team2_home_city'] = ipl_df['team2'].map(team_cities)

print(ipl_df[['team1', 'team1_home_city', 'team2', 'team2_home_city']].head())
```

```python
pivot_wins = pd.pivot_table(
    ipl_df,
    index='season',
    columns='winner',
    values='id',
    aggfunc='count',
    fill_value=0
)

print(pivot_wins)
```

```python
top_3 = pivot_wins.apply(lambda row: row.nlargest(3), axis=1)

print(top_3)
```
